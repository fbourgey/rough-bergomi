<font color="orange">Weak approximations and VIX option price
expansions in forward variance curve models</font>

Convention is different. The parameter $\eta$ in the paper must be 
replaced with $\eta \sqrt{2H}$

Given a kernel function $K(\cdot)$, consider the SDE for the instantaneous 
forward variance curve model
$$
    d\xi_t^u = \xi_t^u K(u,t) \, dW_t
$$
or equivalently, for all $u \geq T$,
$$
    \xi_T^u = \xi_0^u \exp \left\{ Z_T^u \right\}
$$
where 
$$
    Z_T^u = -\frac{1}{2}\int_0^T K(u,t)^2 \, dt + \int_0^T K(u,t) \, dW_t.
$$

We approximte the 

$$
    \text{VIX}_T^2 
    = \frac{1}{\Delta} \int_{T}^{T+\Delta} \xi_T^u \, du
    = \frac{1}{\Delta} \int_{T}^{T+\Delta} \xi_0^u \exp \{Z_T^u\} \, du
$$

with a lognormal proxy

$$
    \text{VIX}_{T,P}^2 = \mathbb{E}[\text{VIX}_T^2]
    \exp \left\{\frac{1}{\Delta} \int_{T}^{T+\Delta} \xi_0^u Z_T^u 
    \frac{du}{\mathbb{E}[\text{VIX}_T^2]}\right\}
$$

where the fair price of a VIX squared futures contract is given by

$$
    \mathbb{E}[\text{VIX}_T^2] = \frac{1}{\Delta} 
    \int_{T}^{T+\Delta} \xi_0^u du.
$$

We have that $\log \text{VIX}_{T,P}^2$ is normal with mean

$$
    \mu_{P} = 
    \log \mathbb{E}[\text{VIX}_T^2] -
    \frac12 \int_0^T \left(\frac1\Delta 
    \int_{T}^{T+\Delta} \xi_0^u K(u,t)^2 \frac{du}{\mathbb{E}[\text{VIX}_T^2]} \right) \, dt
$$

and variance

$$
    \sigma_{P}^2 = 
    \int_0^T \left(\frac1\Delta 
    \int_{T}^{T+\Delta} \xi_0^u K(u,t) \frac{du}{\mathbb{E}[\text{VIX}_T^2]} \right)^2 \, dt
$$

---

We approximate the VIX option price by
$$
    \mathbb{E}[\varphi(\text{VIX}_{T}^2)] \approx
    \mathbb{E}[\varphi(\text{VIX}_{T,P}^2)]
    + \sum_{i=1}^{3} \gamma_i \, \partial_{\varepsilon}\left.\mathbb{E}\left[\varphi(\text{VIX}_{T,P}^2 \, e^{\varepsilon})\right]\right|_{\varepsilon=0}
$$

- The zeroth order approximation corresponds to a Black-Scholes price
while the other terms are corrections based on Black-Scholes Greeks (delta, gamma, speed).

- The coefficients $\gamma_i$ are deterministic integrals depending on the kernel.

## Price of VIX futures wrt the vol-of-variance $\eta$

```python
tab_eta = np.linspace(0.1, 1.5, 10) * np.sqrt(2.0 * params["H"])

# Monte Carlo price
vix_fut_eta = {}
for month in mats:
    print(month)
    price = np.zeros_like(tab_eta)
    for i, eta in enumerate(tab_eta):
        rbergomi.eta = eta
        price[i] = rbergomi.price_vix_fut(
            T=mats[month], n_disc=n_disc, n_mc=n_mc, seed=SEED
        )
    vix_fut_eta[month] = price
```

# Misc 

Try to approximate the integral

$$
    \int_{-\infty}^{B} (1 + b e ^{c y})^{1/2} \, \phi(y) \, dy
$$

```python
def integral(b, c, B):
    return integrate.quad(
        lambda y: np.sqrt(1 + b * np.exp(c * y)) * stats.norm.pdf(y), -np.inf, B
    )[0]


def integral_approx_small(b, c, B, order=3):
    # small expansion: |b * np.exp(c * B)| << 1
    term0 = stats.norm.cdf(B)
    if order == 0:
        return term0
    term1 = (b / 2.0) * np.exp(c**2 / 2.0) * stats.norm.cdf(B - c)
    if order == 1:
        return term0 + term1
    term2 = (-(b**2) / 8.0) * np.exp(2 * c**2) * stats.norm.cdf(B - 2 * c)
    if order == 2:
        return term0 + term1 + term2
    term3 = (b**3 / 16.0) * np.exp(9 * c**2 / 2.0) * stats.norm.cdf(B - 3 * c)
    if order == 3:
        return term0 + term1 + term2 + term3

    raise ValueError("order must be in {0,1,2,3}")


def integral_approx_large(b, c, B, order=3):
    # large expansion: |b * np.exp(c * B)| >> 1
    term0 = np.sqrt(b) * np.exp(c**2 / 8.0) * stats.norm.cdf(B - c / 2)
    if order == 0:
        return term0
    term1 = (1.0 / (2.0 * np.sqrt(b))) * np.exp(c**2 / 8.0) * stats.norm.cdf(B + c / 2)
    if order == 1:
        return term0 + term1
```

```python
print("Numerical vs Approx integral")
b = 0.2
c = 1.0
B = 10
print("b =", b, "c =", c, "B =", B)
print(f"b * exp(c * B) = {b * np.exp(c * B)}")
print(f"Numerical: {integral(b, c, B)}")
print(f"Approx large b exp(c B): {integral_approx_large(b, c, B, order=1)}")
print(f"Approx small b exp(c B): {integral_approx_small(b, c, B, order=0)}")
```
